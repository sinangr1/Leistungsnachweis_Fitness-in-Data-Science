# Kalorie ausberechnung in Running der Menschen, die zwischen 60 und 70 jahre alte sind:
# dff_run_age_6070

```python
dff_run_age_6070 = dff[(dff['Age'] >= 60) & (dff['Age'] <= 70)]
dff_run_age_6070['Age'].value_counts()
```

```python
run_steps_mean_6070 = dff_run_age_6070['Steps'].mean()


run_Calories_Burned_mean_6070 = dff_run_age_6070['Calories_Burned'].mean()


circa_in_100_steps_6070 = (run_steps_mean_6070*100)/run_Calories_Burned_mean_6070


run_steps_mean_6070, run_Calories_Burned_mean_6070, circa_in_100_steps_6070

# Es zeigt im Unten dass die Menschen, die zwischen 60 und 70 Jahre alte sind geben 2000 Kalorie in 100 steps aus.
```

# Analysen zu Rauchern und Nichtrauchern, zu Erkrankten und Nicht-Erkrankten

```python
df['Smoker'].value_counts()
```

```python
# Analyse von Rauchern

Raucher_Calories_Burned = df[(df['Smoker'] == 'Yes') & (df['Exercise_Type'] == 'Running')]['Calories_Burned'].mean()
Raucher_Calories_Burned
```

```python
Nichtraucher_Calories_Burned=df[(df['Smoker'] == 'No') & (df['Exercise_Type'] == 'Running')]['Calories_Burned'].mean()
Nichtraucher_Calories_Burned
```

```python
Rauchern_Stepszahlen = df[(df['Smoker'] == 'Yes') & (df['Exercise_Type'] == 'Running')]['Steps'].mean()
Rauchern_Stepszahlen
```

```python
Nichtrauchern_Stepszahlen = df[(df['Smoker'] == 'No') & (df['Exercise_Type'] == 'Running')]['Steps'].mean()
Nichtrauchern_Stepszahlen
```

```python
import matplotlib.pyplot as plt


fig, axs = plt.subplots(1, 2, figsize=(12, 6))  # 1 satır, 2 sütun

categories = ['Raucher', 'Nichtraucher']
calories_burned = [Raucher_Calories_Burned, Nichtraucher_Calories_Burned]

axs[0].bar(categories, calories_burned, color=['red', 'orange'], alpha=0.7)
axs[0].set_ylabel('Calories Burned')
axs[0].set_title('Calories Burned by Raucher and Nichtraucher during Running')

Stepzahlen = [Rauchern_Stepszahlen, Nichtrauchern_Stepszahlen]

axs[1].bar(categories, Stepzahlen, color=['blue', 'green'], alpha=0.7)
axs[1].set_ylabel('Stepzahlen')
axs[1].set_title('Stepzahlen by Raucher and Nichtraucher during Running')

plt.tight_layout()
plt.show()
```
