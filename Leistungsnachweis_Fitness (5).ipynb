# Calorie calculation in separation of people who are between 60 and 70 years old
# dff_run_age_6070

```python
dff_run_age_6070 = dff[(dff['Age'] >= 60) & (dff['Age'] <= 70)]
dff_run_age_6070['Age'].value_counts()
```

```python
run_steps_mean_6070 = dff_run_age_6070['Steps'].mean()


run_Calories_Burned_mean_6070 = dff_run_age_6070['Calories_Burned'].mean()


circa_in_100_steps_6070 = (run_steps_mean_6070*100)/run_Calories_Burned_mean_6070


run_steps_mean_6070, run_Calories_Burned_mean_6070, circa_in_100_steps_6070

# Es zeigt im Unten dass die Menschen, die zwischen 60 und 70 Jahre alte sind geben 2000 Kalorie in 100 steps aus.
```

```python

Raucher_Calories_Burned = dff[(dff['Smoker'] == 'Yes') & (dff['Exercise_Type'] == 'Running')]['Calories_Burned'].mean()
Raucher_Calories_Burned
```

```python
import matplotlib.pyplot as plt


categories = ['Smoker', 'Non smoker']
calories_burned = [Raucher_Calories_Burned, Nichtraucher_Calories_Burned]
stepszahlen = [Rauchern_Stepszahlen, Nichtrauchern_Stepszahlen]

fig, axs = plt.subplots(1, 2, figsize=(12, 6))

axs[0].bar(categories, calories_burned, color=['red', 'orange'], alpha=0.7)
axs[0].set_ylabel('Calories Burned')
axs[0].set_title('Average calories burned during running')

axs[1].bar(categories, stepszahlen, color=['red', 'orange'], alpha=0.7)
axs[1].set_ylabel('Number of steps')
axs[1].set_title('Average daily step count')

plt.tight_layout()
plt.show()
```
